```python
# {"signo": "positivo", "tipo": "impar"}
def analizar_numero(n):
    signo = ""
    tipo = ""
    signo = "positivo" if n > 0 else "negativo" if n < 0 else "cero"
    tipo = "par" if n % 2 == 0 else "impar"
    return {"signo": signo, "tipo": tipo}

print(analizar_numero(5))
print(analizar_numero(-4))
print(analizar_numero(0))
    
```
